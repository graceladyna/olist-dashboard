## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

#### Fixing missing value, invalid value & outlier pada orders (batasi periode & status)

```python
MULAI, AKHIR = pd.Timestamp("2017-01-01"), pd.Timestamp("2018-08-31 23:59:59")

orders_clean = orders[
    (orders.order_status == "delivered")
    & orders.order_delivered_customer_date.notna()
    & orders.order_purchase_timestamp.between(MULAI, AKHIR)
].copy()

orders_clean["delivery_days"] = (
    (orders_clean.order_delivered_customer_date - orders_clean.order_purchase_timestamp)
    .dt.total_seconds() / 86400
).round(1)
orders_clean["is_late"] = (orders_clean.order_delivered_customer_date
                           > orders_clean.order_estimated_delivery_date).astype(int)

print(f"orders: {len(orders):,} -> orders_clean: {len(orders_clean):,} baris")
print(f"Persentase order terlambat: {orders_clean.is_late.mean():.1%}")
print(f"Order dengan delivery_days > 60 (dipertahankan): {(orders_clean.delivery_days > 60).sum()}")
```

#### Fixing duplicate data & invalid value pada geolocation

```python
geo = geolocation.drop_duplicates().copy()
geo = geo[geo.geolocation_lat.between(-34, 6) & geo.geolocation_lng.between(-74, -34)]

# Samakan penulisan nama kota: hilangkan aksen, huruf kecil
geo["geolocation_city"] = (geo.geolocation_city.str.normalize("NFKD")
                           .str.encode("ascii", errors="ignore").str.decode("ascii")
                           .str.lower().str.strip())

# Satu titik per kode pos (rata-rata koordinat)
geo_zip = (geo.groupby("geolocation_zip_code_prefix")
              .agg(lat=("geolocation_lat", "mean"), lng=("geolocation_lng", "mean"))
              .reset_index())
print(f"geolocation: {len(geolocation):,} -> {len(geo):,} baris -> {len(geo_zip):,} kode pos unik")
```

#### Fixing duplicate data pada reviews

```python
# Jika satu order punya beberapa review, ambil review terbaru
reviews_clean = (reviews.sort_values("review_answer_timestamp")
                        .drop_duplicates("order_id", keep="last")[["order_id", "review_score"]])
print(f"reviews: {len(reviews):,} -> {len(reviews_clean):,} baris")
```

#### Fixing missing & inconsistent value pada products

```python
terjemahan_manual = {"pc_gamer": "pc_gamer",
                     "portateis_cozinha_e_preparadores_de_alimentos": "portable_kitchen_food_preparers"}

products_clean = products.merge(categories, on="product_category_name", how="left")
products_clean["category"] = (products_clean.product_category_name_english
                              .fillna(products_clean.product_category_name.map(terjemahan_manual))
                              .fillna("unknown"))
print("Produk tanpa kategori ('unknown'):", (products_clean.category == "unknown").sum())
```

#### Menggabungkan tabel menjadi satu dataset utama (main_data)

```python
main = (order_items
        .merge(orders_clean[["order_id", "customer_id", "order_purchase_timestamp",
                             "delivery_days", "is_late"]], on="order_id", how="inner")
        .merge(customers[["customer_id", "customer_unique_id", "customer_zip_code_prefix", "customer_state"]],
               on="customer_id", how="left")
        .merge(products_clean[["product_id", "category"]], on="product_id", how="left")
        .merge(reviews_clean, on="order_id", how="left")
        .merge(geo_zip, left_on="customer_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left"))
main["category"] = main.category.fillna("unknown")

kolom = ["order_id", "order_item_id", "customer_unique_id", "customer_state", "order_purchase_timestamp",
         "category", "price", "freight_value", "review_score", "delivery_days", "is_late", "lat", "lng"]
main = main[kolom]

print("Bentuk main_data:", main.shape)
print("Missing value tersisa:\n" + main.isna().sum()[main.isna().sum() > 0].to_string())

# Simpan untuk dashboard (buat folder dashboard/ lebih dulu)
out = main.copy()
out["lat"], out["lng"] = out.lat.round(4), out.lng.round(4)
out.to_csv("dashboard/main_data.csv", index=False)
main.head()
```

**Insight:**
- Setelah pembersihan dan penggabungan, `main_data` berisi **109.872 baris item** dari **96.203 order** dan **93.096 pelanggan unik**.
- Sisa missing value wajar: 823 baris tanpa review dan 289 baris tanpa koordinat. Baris ini tetap dipakai kecuali untuk analisis yang membutuhkan kolom tersebut.
- Sekitar **8,1%** order tiba setelah tanggal estimasi (`is_late = 1`).

## Visualization & Explanatory Analysis

### Pertanyaan 1: Kategori produk apa yang menyumbang revenue tertinggi, dan mana yang review-nya di bawah rata-rata?

```python
kat = (main.groupby("category")
           .agg(revenue=("price", "sum"), item_terjual=("order_item_id", "count"), jumlah_order=("order_id", "nunique"))
           .sort_values("revenue", ascending=False))
kat["pct_revenue"] = kat.revenue / kat.revenue.sum() * 100
kat["pct_kumulatif"] = kat.pct_revenue.cumsum()
kat["avg_harga_item"] = kat.revenue / kat.item_terjual

rev_kat = main.drop_duplicates(["order_id", "category"]).groupby("category").review_score.mean()
kat["avg_review"] = rev_kat

top10 = kat.head(10)
print(f"10 kategori teratas menyumbang {top10.pct_revenue.sum():.1f}% dari total revenue ({len(kat)} kategori).")
top10
```

```python
top10 = kat.head(10).iloc[::-1]          # dibalik agar kategori terbesar tampil paling atas
warna = [HIGHLIGHT if i >= 7 else MUTED for i in range(10)]   # sorot 3 teratas

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5), sharey=True, gridspec_kw={"width_ratios": [3, 1.4]})

ax1.barh(top10.index, top10.revenue, color=warna)
for y, (rev, pct) in enumerate(zip(top10.revenue, top10.pct_revenue)):
    ax1.text(rev + top10.revenue.max() * 0.01, y, f"R$ {rev/1e6:.2f} jt ({pct:.1f}%)", va="center", fontsize=9)
ax1.set_xlim(0, top10.revenue.max() * 1.28)
ax1.xaxis.set_major_formatter(FuncFormatter(rupiah_juta))
ax1.set_title("Revenue per kategori (10 teratas), Jan 2017 - Agu 2018", loc="left")
ax1.set_xlabel("Revenue dari harga item (R$)")
ax1.grid(axis="y", visible=False)

rata_review = order_lvl.review_score.mean()
# dot plot (bukan bar) karena sumbu tidak dimulai dari 0
ax2.hlines(top10.index, rata_review, top10.avg_review, color=MUTED, lw=2)
ax2.scatter(top10.avg_review, top10.index, s=70, zorder=3,
            color=[ALERT if v < rata_review else HIGHLIGHT for v in top10.avg_review])
for y, v in enumerate(top10.avg_review):
    ax2.text(v + (0.015 if v >= rata_review else -0.015), y, f"{v:.2f}", va="center",
             ha="left" if v >= rata_review else "right", fontsize=9)
ax2.axvline(rata_review, color="black", ls="--", lw=1)
ax2.set_xlim(3.9, 4.35)
ax2.set_title("Rata-rata review (1-5)", loc="left")
ax2.set_xlabel(f"Skor review (-- rata-rata semua order: {rata_review:.2f})")
ax2.grid(axis="y", visible=False)
sns.despine()
plt.tight_layout()
plt.show()
```

### Analisis Lanjutan 1: RFM Analysis
**Tujuan:** mengelompokkan pelanggan berdasarkan perilaku belanja agar strategi retensi bisa dibedakan per kelompok.
- **Recency:** hari sejak pembelian terakhir. **Frequency:** jumlah order per pelanggan. **Monetary:** total belanja.

Skor R dan M memakai **kuartil** (4 = terbaik). Skor F memakai **aturan bisnis** (1, 2, >=3 order) karena sebagian besar pelanggan hanya membeli sekali. Segmen dibuat dengan *manual grouping*.

```python
snapshot = main.order_purchase_timestamp.max() + pd.Timedelta(days=1)

rfm = (main.groupby("customer_unique_id")
           .agg(terakhir_beli=("order_purchase_timestamp", "max"),
                frequency=("order_id", "nunique"),
                monetary=("price", "sum"))
           .reset_index())
rfm["recency"] = (snapshot - rfm.terakhir_beli).dt.days

print("Distribusi frequency (jumlah order per pelanggan):")
print(rfm.frequency.value_counts().sort_index().head(6).to_string())
print(f"\nPelanggan yang hanya membeli 1x: {(rfm.frequency == 1).mean():.1%}")

rfm["R"] = pd.qcut(rfm.recency.rank(method="first"), 4, labels=[4, 3, 2, 1]).astype(int)
rfm["M"] = pd.qcut(rfm.monetary.rank(method="first"), 4, labels=[1, 2, 3, 4]).astype(int)
rfm["F"] = np.select([rfm.frequency >= 3, rfm.frequency == 2], [3, 2], default=1)

rfm["segmen"] = np.select(
    [(rfm.F >= 2) & (rfm.R >= 3) & (rfm.M >= 3),
     (rfm.F >= 2),
     (rfm.R >= 3) & (rfm.M >= 3),
     (rfm.R >= 3),
     (rfm.M >= 3)],
    ["Champions", "Repeat Buyers", "Recent Big Spenders", "Recent One-time", "Lapsed Big Spenders"],
    default="Lapsed One-time")

urutan = ["Champions", "Repeat Buyers", "Recent Big Spenders", "Recent One-time", "Lapsed Big Spenders", "Lapsed One-time"]
seg = (rfm.groupby("segmen").agg(pelanggan=("customer_unique_id", "count"), avg_recency=("recency", "mean"),
                                 avg_frequency=("frequency", "mean"), avg_monetary=("monetary", "mean"),
                                 total_revenue=("monetary", "sum")).reindex(urutan))
seg["pct_pelanggan"] = seg.pelanggan / seg.pelanggan.sum() * 100
seg["pct_revenue"] = seg.total_revenue / seg.total_revenue.sum() * 100
seg
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4.8), sharey=True)
y = seg.index[::-1]
ax1.barh(y, seg.pct_pelanggan[::-1], color=MUTED)
ax2.barh(y, seg.pct_revenue[::-1], color=HIGHLIGHT)
for ax, col, ttl in ((ax1, seg.pct_pelanggan, "Porsi pelanggan (%)"), (ax2, seg.pct_revenue, "Porsi revenue (%)")):
    for i, v in enumerate(col[::-1]):
        ax.text(v + 0.5, i, f"{v:.1f}%", va="center", fontsize=9)
    ax.set_title(ttl, loc="left"); ax.set_xlim(0, max(seg.pct_pelanggan.max(), seg.pct_revenue.max()) * 1.18)
    ax.grid(axis="y", visible=False)
fig.suptitle("Segmentasi pelanggan (RFM): porsi pelanggan vs porsi revenue", x=0.01, ha="left", fontsize=14, fontweight="bold")
sns.despine(); plt.tight_layout(rect=[0, 0, 1, 0.95]); plt.show()
```

**Insight:**
- **97,0%** pelanggan hanya membeli **sekali**. Pelanggan berulang (*Champions* + *Repeat Buyers*) hanya **3,0%** pelanggan dan **5,5%** revenue; retensi adalah masalah terbesar.
- Segmen *Big Spenders* menyumbang **78%** revenue dari sekitar **47,5%** pelanggan. Separuhnya (*Lapsed Big Spenders*: 23,6% pelanggan, 38,9% revenue) sudah lama tidak kembali, kelompok bernilai tinggi yang paling layak dihubungi ulang.
- Pelanggan sekali-beli berbelanja kecil (rata-rata sekitar R$ 47).
